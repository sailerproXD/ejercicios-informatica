```python
import math
# Elaborar un algortimo que me de el area y perimetro de un circuferencia 
R = float(input("ingrese el radio del circulo: "))
Ac = math.pi*(R**2)
Pc = math.pi*2*R
print(f'El area del circulo es: {Ac:.2f}\nEl perimetro del circulo es: {Pc:.2f}')
```

```python
# Dado el numero de pasos determianr cuantos kilometro, metros y cetimetros avanza un apersona
P = float(input("ingrese el numero de pasos: "))
CM = P*45
M = CM/100
Km = M/1000
print(f'Numero de centimetros avanzados: {CM}\nNumero de metros avanzados:{M}\nNumero de kilometros avanzados:{Km}')
```

```python
import math
#Realizar un algoritmo que convierta de grados a radianes 
Numero = float(input("ingrese el numero de grados: "))
Radianes = Numero*math.pi/180
print(f'conversion de grados a radianes es: {Radianes:.2f}')
```

```python
#Realizar un algoritmo que convierta de grados celcuis a kelvin, y kelvin a Fahreenheit
C = float(input("ingrese los grados celcius: "))
K = C + 273
F = ( C*1.8) + 32
print(f'Conversion de grados celcius a kelvin: {K:.3f}\nConversion de grados celcius a Fahreenheit: {F:.3f}')
```
