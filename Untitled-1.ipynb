# Algoritmo que mne da el promedio de 5 notas, escribir si su calificacion es aprobada o igual a 3, mostrar promedio

```python
n1 = float(input("ingrese la primera nota:"))*0.30
n2 = float(input("ingrese la segunda nota:"))*0.15
n3 = float(input("ingrese la tercera nota:"))*0.15
n4 = float(input("ingrese la cuarta nota:"))*0.20
n5 = float(input("ingrese la quinta nota:"))*0.20
Nota = (n1 + n2 + n3 + n4 + n5)
if Nota >= 3.0:
  print("aprobado")
else:
  print("reprobado")
print( " promedio del estudiante es: {x}".format(x=Nota))
```
